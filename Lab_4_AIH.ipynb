```python
# Calculate outliers using IQR

Q1 = df.drop(columns='Outcome').quantile(0.25)
Q3 = df.drop(columns='Outcome').quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = ((df.drop(columns='Outcome') < lower_bound) |
            (df.drop(columns='Outcome') > upper_bound))

outlier_count = outliers.sum()

print("Number of outliers in each feature:")
print(outlier_count)
```

```python
# Features
X = df.drop('Outcome', axis=1)

# Target
y = df['Outcome']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
print("Training class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit only on training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data
X_test_scaled = scaler.transform(X_test)
```

```python
print("First 5 rows of standardized training data:")
print(X_train_scaled[:5])
```

```python
from sklearn.linear_model import LogisticRegression

# Create model
model = LogisticRegression(max_iter=1000)

# Train model
model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully.")
```

```python
# Predict class labels
y_pred = model.predict(X_test_scaled)

print("Predicted values:")
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    target_names=['Malignant', 'Benign']
))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Malignant', 'Benign'],
    yticklabels=['Malignant', 'Benign']
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix")

plt.show()
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

# Probability of positive class
y_prob = model.predict_proba(X_test_scaled)[:, 1]

# Calculate ROC values
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# Calculate AUC
auc_score = roc_auc_score(y_test, y_prob)

print("AUC Score:", auc_score)
```

```python
plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    label=f'Logistic Regression (AUC = {auc_score:.3f})'
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--',
    label='Random Classifier'
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")

plt.legend()

plt.show()
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy = accuracy_score(y_test, y_pred)

precision_malignant = precision_score(y_test, y_pred, pos_label=0)
recall_malignant = recall_score(y_test, y_pred, pos_label=0)
f1_malignant = f1_score(y_test, y_pred, pos_label=0)

precision_benign = precision_score(y_test, y_pred, pos_label=1)
recall_benign = recall_score(y_test, y_pred, pos_label=1)
f1_benign = f1_score(y_test, y_pred, pos_label=1)

auc = roc_auc_score(y_test, y_prob)

print("========== MODEL EVALUATION ==========")

print(f"Accuracy: {accuracy:.4f}")

print("\nMalignant Class:")
print(f"Precision: {precision_malignant:.4f}")
print(f"Recall:    {recall_malignant:.4f}")
print(f"F1-score:  {f1_malignant:.4f}")

print("\nBenign Class:")
print(f"Precision: {precision_benign:.4f}")
print(f"Recall:    {recall_benign:.4f}")
print(f"F1-score:  {f1_benign:.4f}")

print(f"\nAUC: {auc:.4f}")
```
